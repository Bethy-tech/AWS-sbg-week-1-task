# Student Data Cleaning Notebook

This notebook loads the `students.csv` file from the same folder as this notebook, cleans the data using reusable functions from `task.py`, and computes summary statistics with NumPy.


```python
from pathlib import Path
import sys

repo_root = Path.cwd()
week_dir = repo_root / 'week-01'
csv_candidates = [
    Path('students.csv'),
    repo_root / 'students.csv',
    week_dir / 'students.csv',
]

data_path = next((p for p in csv_candidates if p.exists()), week_dir / 'students.csv')
if not data_path.exists():
    raise FileNotFoundError(f'Could not find students.csv. Checked: {csv_candidates}')

print(f'Dataset path: {data_path}')

if week_dir.exists() and str(week_dir) not in sys.path:
    sys.path.insert(0, str(week_dir))

from task import load_csv_dataset

raw_records = load_csv_dataset(data_path)
print(f'Raw records loaded: {len(raw_records)}')
```

## Data cleaning with functions

The project defines functions for standardizing names, converting values to numeric types, filling missing data, removing duplicates, and replacing invalid ages or scores. This notebook applies the full cleaning workflow and shows the result in a readable format.


```python
from task import clean_student_data

cleaned_records = clean_student_data(raw_records)
print(f'Cleaned records: {len(cleaned_records)}')
for index, record in enumerate(cleaned_records, start=1):
    print(f'{index}. {record["name"]} | age={float(record["age"]):.0f} | score={float(record["score"]):.2f}')
```

## NumPy summary statistics

After cleaning the records, we convert the score column into a NumPy array and compute the mean, median, standard deviation, percentiles, and min/max values. These statistics help summarize the data in a compact, readable way.


```python
import numpy as np
from task import calculate_statistics

scores = np.array([float(record['score']) for record in cleaned_records], dtype=float)
stats = calculate_statistics(cleaned_records)

print('\nSummary statistics for student scores:')
print(f"Mean score: {stats['mean']:.2f}")
print(f"Median score: {stats['median']:.2f}")
print(f"Standard deviation: {stats['std_dev']:.2f}")
print(f"90th percentile: {stats['p90']:.2f}")
print(f"Minimum score: {stats['min']:.2f}")
print(f"Maximum score: {stats['max']:.2f}")
```
